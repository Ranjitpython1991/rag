RAG (Retrieval-Augmented Generation) is a technique that enhances the performance of language models by combining them with an external retrieval mechanism. This allows the model to pull in relevant information from large document stores or knowledge bases at inference time, improving the quality and factual accuracy of its generated responses.

Indexing is the process of converting raw data into a searchable format (embeddings) 
and storing it in a vector database for fast retrieval.

Load: First we need to load our data. This is done with Document Loaders.

Split: Text splitters break large Documents into smaller chunks. This is useful both for indexing data and passing it into a model, as large chunks are harder to search over and won’t fit in a model’s finite context window.

Store: We need somewhere to store and index our splits, so that they can be searched over later. This is often done using a VectorStore and Embeddings model.

Context window = the maximum amount of text (in tokens) an LLM can read at one time.

If separators are not specified, RecursiveCharacterTextSplitter uses a default hierarchy like paragraphs, lines, spaces, and finally character-level splitting to ensure chunks stay within size limits.

Vector store

A vector stores embedded data and performs similarity search.

add_documents - Add documents to the store.

delete - Remove stored documents by ID.

similarity_search - Query for semantically similar documents.

 semantic query using similarity_search, which returns the closest embedded documents

 Many vector stores support parameters like:

k — number of results to return
filter — conditional filtering based on metadata


vector_store.similarity_search(
  "query",
  k=3,
  filter={"source": "tweets"}
)
